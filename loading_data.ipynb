```python
from sqlalchemy import create_engine
import datetime

engine = create_engine('sqlite:///db.sqlite3', echo=False)


data_product['is_active'] = True
data_product['created_at'] = datetime.datetime.now()
data_product['updated_at'] = datetime.datetime.now()

data_category['created_at'] = datetime.datetime.now()
data_category['updated_at'] = datetime.datetime.now()

data_rating['created_at'] = datetime.datetime.now()


data_product.to_sql(
    name='store_product',       # Tên bảng trong database (app: store, model: Product)
    con=engine,                 
    if_exists='replace',         # Thêm dòng mới, giữ nguyên cấu trúc bảng
    # index=False                 # Bỏ qua cột index của DataFrame
)

data_category.to_sql(
    name='store_category',       # Tên bảng trong database (app: store, model: Category)
    con=engine,                 
    if_exists='replace',         # Thêm dòng mới, giữ nguyên cấu trúc bảng
    # index=False                 # Bỏ qua cột index của DataFrame
)

data_rating.to_sql(
    name='store_rating',       # Tên bảng trong database (app: store, model: Rating)
    con=engine,                 
    if_exists='replace',         # Thêm dòng mới, giữ nguyên cấu trúc bảng
    index=False                 # Bỏ qua cột index của DataFrame
)

print("Đã đẩy dữ liệu vào bảng store_product, store_category, store_rating trong SQLite thành công!")
```
